```python
#DFS를 이용한 친구추천
from collections import deque
graph = {
  'Alice':['Bob','Charlie'],
  'Bob':['David','Eve'],
  'Charlie':['Frank'],
  'David':[],
  'Eve':['Frank'],
  'Frank':[]
}

#1. DFS 사용
def dfs(graph,v,visited):
  visited.add(v)
  print(v,end=' ')
  for neighbor in graph[v]:
    if neighbor not in visited:
      dfs(graph,neighbor,visited)
print("DFS 탐색순서")
dfs(graph,'Alice',set())
```

```python
#BFS를 사용한 친구추천
from collections import deque
graph = {
  'Alice':['Bob','Charlie'],
  'Bob':['David','Eve'],
  'Charlie':['Frank'],
  'David':[],
  'Eve':['Frank'],
  'Frank':[]
}

#1. BFS사용
def bfs(graph,start):
  visited = set()
  queue = deque([start])
  visited.add(start)
  while queue:
    w = queue.popleft()
    print(w,end=' ')
    for neighbor in graph[w]:
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)
print("BFS 탐색 순서")
bfs(graph,'Alice')
```

```python
#BFS를 이용한 친구추천 시스템
from collections import deque
graph = {
  'Alice':['Bob','Charlie'],
  'Bob':['David','Eve'],
  'Charlie':['Frank'],
  'David':[],
  'Eve':['Frank'],
  'Frank':[]
}

def recommend_friend(graph,start):
  visited = set()
  queue = deque([(start,0)])
  print(f"\n{start}의 친구 추천 (2단계):")
  while queue:
    current, depth = queue.popleft()
    if current not in visited:
      visited.add(current)
      if depth == 2:
        print("-",current)
      if depth < 2:
        for neighbor in graph[current]:
          if neighbor not in visited:
            queue.append((neighbor,depth+1))
recommend_friend(graph,'Alice')
```
